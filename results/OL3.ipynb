### Cifar10

```python
results1 = np.loadtxt("cifar10-single-acc-results.txt")
results2 = np.loadtxt("cifar10-wide-acc-results.txt")
results3 = np.loadtxt("cifar10-deep-acc-results.txt")
results4 = np.loadtxt("cifar10-resid-acc-results.txt")
results5 = np.loadtxt("cifar10-custom")

plt.plot(np.arange(0,results1.shape[1]),
         np.mean(results1,0),label="Single")
plt.fill_between(np.arange(0,results1.shape[1]),
                 np.mean(results1,0)-(1.96*np.std(results1,0)/np.sqrt(results1.shape[0])),
                 np.mean(results1,0)+(1.96*np.std(results1,0)/np.sqrt(results1.shape[0])),
                 alpha=0.5)

plt.plot(np.arange(0,results2.shape[1]),
         np.mean(results2,0),label="Wide")
plt.fill_between(np.arange(0,results2.shape[1]),
                 np.mean(results2,0)-(1.96*np.std(results2,0)/np.sqrt(results2.shape[0])),
                 np.mean(results2,0)+(1.96*np.std(results2,0)/np.sqrt(results2.shape[0])),
                 alpha=0.5)

plt.plot(np.arange(0,results3.shape[1]),
         np.mean(results3,0),label="Deep")
plt.fill_between(np.arange(0,results3.shape[1]),
                 np.mean(results3,0)-(1.96*np.std(results3,0)/np.sqrt(results3.shape[0])),
                 np.mean(results3,0)+(1.96*np.std(results3,0)/np.sqrt(results3.shape[0])),
                 alpha=0.5)

plt.plot(np.arange(0,results4.shape[1]),
         np.mean(results4,0),label="Residual")
plt.fill_between(np.arange(0,results4.shape[1]),
                 np.mean(results4,0)-(1.96*np.std(results4,0)/np.sqrt(results4.shape[0])),
                 np.mean(results4,0)+(1.96*np.std(results4,0)/np.sqrt(results4.shape[0])),
                 alpha=0.5)

plt.ylim([0.0,0.6])
plt.ylabel('Validation Accuracy (P)')
plt.xlabel('Epoch')
plt.legend(loc='lower right')
plt.show()
```

### Reflection

### 1. Which of the architectures seems to perform best overall even under the constraints above? Why do you think this may be happening?
The residual architecture seems to do the best out of all of them under the constraints, peaking at 50.35% on Cifar100. This is likely because of the architectural difference itself, where gradients are allowed to skip layers through residual connections. They make better use of the parameter budget.

### 2. Considering the architectural differences between single, wide, deep, and residual networks, what other structural choices or modifications do you think might further improve performance under the parameter constraints given?
I think maybe making the layers deeper in certain circumstances might help, as more layers allows the model to represent more complex functions. Also, changing the activation function to ReLU might help with performance, fighting vanishing gradients that Tanh functions are known to suffer from. Also, using convolution layers would help, as it would be a better method of tackling a task like this.

### 3. What outcome did you expect from your custom modification (hypothesis), and did your results confirm or challenge that expectation?
I expected the accuracy to go up for every single architecture, however, that didn't happen. It challenged that hypothesis, however, horizontal flipping did act as a regularizer against overfitting. For instance, in the wide architecture, loss was down by 0.4, while Residual went down by 0.37, Deep went down by 0.1, and Single remained about the same.

### 4. What process would you use to more rigorously confirm your hypothesis and what steps would testing it involve?
I would perform more seeds, and compare the ending accuracy and loss between custom and non-custom architectures. Getting more samples allows me to be more confident in whether my hypothesis is right or wrong. Performing an independent t-test comparing all the runs and checking the resulting p-value would establish statistic significance if it is lower than 0.05.
